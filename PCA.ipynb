```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
```

```python
iris = load_iris()

X = pd.DataFrame(iris.data)
y = iris.target
```

```python
X.head()
```

```python
# standardize 

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled.shape # original shape of data
```

```python
# PCA

from sklearn.decomposition import PCA


pca = PCA(n_components = 2) # by default the min value among n_features or n_samples is chosen for n_components

X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca.shape # after dimension reduction shape
```

```python
print("explained variance ratio :", pca.explained_variance_ratio_)
```

```python
print(pca.components_) # eigen vector values pca1 and pca2 {row wise}
```

```python
# pc1 - combination of 4 features
# pc2 - combination of 4 features
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    c=y
)

plt.xlabel("pc1")
plt.ylabel("pc2")
plt.show()
```
